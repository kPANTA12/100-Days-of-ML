# NLP with Machine Learning

#### 1. Import Libraries

```python
import pandas as pd
import numpy as np
import nltk
import string

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix
```

#### 2. Sample Dataset

```python
data = {
    "text": [
        "Congratulations! You won a free lottery ticket",
        "Hey, are we meeting tomorrow?",
        "Claim your free prize now",
        "Let's have lunch today",
        "Win cash now!!!",
        "Project meeting scheduled at 5 PM"
    ],
    "label": [1, 0, 1, 0, 1, 0]  # 1 = Spam, 0 = Not Spam
}

df = pd.DataFrame(data)
```

#### 3. Text Cleaning

```python
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = ''.join([char for char in text if char not in string.punctuation])
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return ' '.join(words)

df['cleaned_text'] = df['text'].apply(clean_text)
```

#### 4. TF-IDF Vectorization

```python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['cleaned_text'])
y = df['label']
```

#### 5. Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

#### 6. Train Naive Bayes Model

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

#### 7. Evaluate

```python
predictions = model.predict(X_test)

print(confusion_matrix(y_test, predictions))
print(classification_report(y_test, predictions))
```

#### 8. Test Custom Message

```python
new_message = ["Congratulations you have won free money"]

new_clean = [clean_text(new_message[0])]
new_vector = vectorizer.transform(new_clean)

print("Prediction (1=Spam, 0=Not Spam):", model.predict(new_vector))
```
