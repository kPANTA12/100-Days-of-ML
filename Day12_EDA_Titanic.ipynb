# Exploratory Data Analysis (EDA)

### 1. Importing the required libraries for EDA

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 2. Loading the data into the dataframe

```python
df = sns.load_dataset('titanic') # Uploding file predefined in sns
df
```

### 3. Checking the types of Data

```python
df.dtypes
```

### 4. Dropping the irrelevent columns

```python
df = df.drop(['parch'], axis=1)
df
```

### 5. Renaming the columns

```python
df= df.rename(columns = {'survived':'Survived', 'pclass':'Pclass', 'sex':'Sex', 'age':'Age', 'sibsp':'Sibsp', 'fare':'Fare', 'embarked':'Embarked',
                         'class':'Class', 'who':'Who', 'adult_male':'Adult_male', 'deck':'Deck', 'embark_town':'Embark_town', 'alive':'Alive','alone':'Alone'})

                         
```

```python
df
```

### 6. Dropping the Duplicate Rows

```python
df.shape #Checking the total number of rows and columns
```

```python
duplicated_rows_df = df[df.duplicated()]
print("No of duplicated rows: " , duplicated_rows_df.shape)
```

```python
df= df.drop_duplicates()
df
```

### 7. Droping the Missing and Null value

```python
df= df.dropna()
df
```

### 8. Detecting Outliers

```python
sns. boxplot(x= df['Age'])
```

```python
sns.boxplot(x=df['Fare'])
```

### 9. Removing the outliers using IQR method

```python
Q1 = df['Fare'].quantile(0.25)
Q3= df['Fare']. quantile(0.75)
IQR = Q3 -Q1
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df = df.loc[(df['Fare'] >= lower_bound) & (df['Fare'] <= upper_bound)]
df
```

```python
sns.boxplot(x=df['Fare'])
```

### 9. Plot different features against one another (scatter), against frequency (histogram)

#### a. Histogram

```python
plt.figure(figsize=(10,6), dpi=300)
plt.hist(df['Age'], bins=30, color='r', label=" Age")
plt.legend()
plt.xlabel("Age")
plt.ylabel("No of Peoples")
plt.title("Number of People by Age")
```

#### b. Heat Maps

```python
pivot_matrix = df.pivot_table(values='Fare', index='Class', columns='Sex', observed=True)
pivot_matrix
```

```python
sns.heatmap(pivot_matrix, annot=True, cmap='viridis')
plt.show()
```

#### c. Scatter Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived')
plt.show()
```
