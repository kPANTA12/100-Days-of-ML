## Linear Algebra

## 1. Scalars, Vectors, Matrices

```python
scalar= 5
vector = np.array([1,2,3])
matrix= np.array([[1,2], [3,4]])
scalar, vector, matrix
```

## 2. Matrix Addition and Scalar Multiplication
Matrix addition involves adding corresponding entries of two matrices of the same size to create a new matrix.
Scalar multiplication is an operation where a matrix is multiplied by a single number (a scalar), and every entry in the matrix is multiplied by that scalar to create a new matrix. 

```python
A = np.array([[1,2,4],[10, 25, 26]])
B = np.array([[1,5,8], [45, 25, 85]])
print("A+ B = \n", A+B)
print("2* A= \n", 2 *A)
```

## 3. Matrix Multiplication 
Matrix multiplication is an operation that produces a new matrix from two given matrices,
where the number of columns in the first matrix must equal the number of rows in the second.

```python
A= np.array([[1,2 , 3], [1, 2, 5]])
B= np. array([[1,2], [4,5], [6, 5]])
np.matmul(A,B)
```

## 4. Transpose of Matrix (row <--> column)
The transpose of a matrix is a new matrix created by swapping the original matrix's rows with its columns.

```python
A = np.array([[5, 6, 7],[8,5, 9], [4,5,6],[8,9,3]])
A.T
```

## 5. Identity Matrix
An identity matrix is a square matrix with ones on the main diagonal and zeros everywhere else. 

```python
np.identity(3) # Here 3 identity order of square matrix
```

## 6. Matrix Inverse
The inverse of a square matrix is another matrix that, when multiplied by the original matrix, results in the identity matrix.

```python
A = np.array([[1,2,3],[5,6,7],[8,9,10]])
np.linalg.inv(A)
```

## 7. Dot Product
The dot product is a scalar (a single number) that results from multiplying two vectors. It can be calculated by multiplying the 
corresponding components of each vector and summing the results, or by multiplying the magnitudes of the two vectors and the cosine
of the angle between them.

```python
x= np.array([1,2 , 3])
y= np.array([5, 6, 7])
np.dot(x,y)
```

## 8. Norm (Magnitude of vector)
The norm is a function, defined on a vector space, that associates to each vector a measure of its length.

```python
v= np.array([1, 2, 3])
np.linalg.norm(v)            
```

## 9. Eigen Values and Eigenvectors
Eigenvectors are non-zero vectors that, when a linear transformation is applied to them, do not change their direction, only their magnitude.
The eigenvalue is the scalar by which the eigenvector is stretched or compressed.

```python
A = np.array([[2,1],[1,2]])
values, vector = np.linalg.eig(A)
values, vector
```

## Statistics

## 1. Mean, Median, Mode
The mean is the average, found by summing all values and dividing by the count.
The median is the middle value when the data is ordered from smallest to largest. 
The mode is the value that appears most frequently in the dataset. 

```python
data= np. array([10, 20, 30, 25, 45, 50, 55])
mean= np.mean(data)
median=np.median(data)
mode=pd.Series(data).mode()[0]
mean, median, mode
```

## 2. Varience and Standard Deviation
Variance calculates the average of the squared differences from the mean, while standard deviation provides this same measure of spread but in the original units of the data, making it easier to interpret. A higher value for either indicates a greater spread, while a value of zero means all data points are identical


```python
var= np.var(data)
std= np.std(data)
var, std
```

## 3. Correlation and Covariance
Both covariance and correlation measure the relationship between two variables. Covariance measures how variables depend on one another and how a change in one variable may lead to another. Correlation shows how they relate and how one variable may impact the other. 

```python
x= np.array([1, 2, 3, 4, 5])
y= np.array([2, 4, 6, 8, 10])
cov= np.cov(x,y)[0][1]
corr= np.corrcoef(x,y)[0][1]
cov, corr
```

## 4. Normal Distribution 
A normal distribution is a symmetrical, bell-shaped probability distribution where most data points cluster around the mean, and the frequency of values decreases symmetrically as they move away from the center.

```python
data= np.random.normal(50, 10, 100)
plt.hist(data, bins= 30)
plt.title("Normal Distribution")
plt.show()
```

## 5. Outliers Detection (IQR method)
Outlier detection is the process of identifying data points that are significantly different from the rest of a dataset.


```python
q1= np.percentile(data, 25)
q3= np.percentile(data, 75)
iqr= q1-q3
lower= q1-1.5*iqr
upper= q3 +1.5*iqr
outliers = data[(data < lower) | (data > upper)]
outliers[:10]
```
