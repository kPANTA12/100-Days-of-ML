# LTSM & GRU

####  Import Libraries

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
```

#### 2. Create Sequence Data

```python
X = np.array ([
    [1,2,3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6]
])
y = np.array([4, 5, 6, 7])
```

#### 3. Reshape for LSTM

```python
X = X.reshape((X.shape[0], X.shape[1], 1))
```

#### 4. Build LSTM Model

```python
model = models.Sequential([
    layers.Input(shape=(3, 1)),
    layers.LSTM(20),
    layers.Dense(1)
])
```

#### 5. Compile & Train

```python
model.compile(optimizer='adam', loss='mse')
model.fit(X, y, epochs=300, verbose=0)
```

#### 6. Predict

```python
test = np.array([[5, 6, 7]]).reshape((1, 3, 1))
print(model.predict(test))
```

### GRU

```python
model_gru = models.Sequential([
    layers.Input(shape=(3, 1)),
    layers.GRU(20),
    layers.Dense(1)
])

model_gru.compile(optimizer='adam', loss='mse')
model_gru.fit(X, y, epochs=300, verbose=0)

print(model_gru.predict(test))
```
