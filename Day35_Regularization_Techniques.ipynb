# Regularization Techiques

#### 1. Generate Data

```python
import numpy as np
import matplotlib.pyplot as plt
X = np. linspace(-3, 3, 200)
y = X**2 + np.random.normal(0, 1, size=200)
X = X.reshape(-1, 1)
```

#### 2. Train Overfitting Model

```python
import tensorflow as tf
from tensorflow.keras import layers
model = tf.keras.Sequential([
    layers.Dense(128, activation ='relu'),
    layers.Dense(128, activation='relu'),
    layers.Dense(1)
])
model.compile(
    optimizer = 'adam',
    loss ='mse'
)
history  = model.fit(
    X, y,
    epochs= 300,
    validation_split = 0.2,
    verbose = 0
)
```

#### 3. Plot Loss

```python
plt.plot(history.history['loss'], label='Train Loss') # Traning loss
plt.plot(history.history['val_loss'], label='Val Loss') #Validation loss
plt.legend()
plt.show()
```

#### 4. Apply Regularization

```python
from tensorflow.keras import regularizers

model_reg = tf.keras.Sequential([
    layers.Dense(64, activation='relu',
                 kernel_regularizer=regularizers.l2(0.01)),
    layers.Dropout(0.3),
    layers.Dense(1)
])

model_reg.compile(
    optimizer='adam',
    loss='mse'
)

history_reg = model_reg.fit(
    X, y,
    epochs=300,
    validation_split=0.2,
    verbose=0
)
```

#### 5. Compare

```python
plt.plot(history_reg.history['loss'], label='Train Loss')
plt.plot(history_reg.history['val_loss'], label='Val Loss')
plt.legend()
plt.show()
```

Overfitting = memorization

Regularization = controlled learning

L2 → smooth weights

L1 → feature selection

Dropout → independence

Early stopping → stop at peak
