# Linear Regression

## Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

## Create a simple dataset

```python
hours = np.array([2, 3, 4, 5, 6, 7]).reshape(-1, 1)
marks = np.array([50, 55, 60, 65, 70, 75])
hours,marks
```

## Split data into Train + Test

```python
X_train, X_test, y_train, y_test = train_test_split(hours, marks, test_size=0.2, random_state=42)
```

## Create & train (fit) the model

```python
model= LinearRegression()
model.fit(X_train,y_train)
```

## Predict using test data

```python
y_pred = model.predict(X_test)
print("Predicted marks:", y_pred)
```

```python
## Check accuracy (R² Score)
```

```python
accuracy = r2_score(y_test, y_pred)
print("Model R2 Accuracy:", accuracy)
```

## Visualize regression line

```python
plt.scatter(hours, marks)
plt.plot(hours, model.predict(hours), linewidth=2)
plt.xlabel("Hours Studied")
plt.ylabel("Marks Scored")
plt.title("Linear Regression Line")
plt.show()
```
