### Initialize Different Types of Arrays


```python
#All zeros matrix
np.zeros((2,2))
```

```python
#All ones matrix
np.ones((4,2,3))
```

```python
#All other number
np.full((2,2),99)
```

```python
#Random decimal numbers
np.random.rand(2,2)
```

```python
#Random integer matrix
np.random.randint(7,size=(3,3))
```

```python
#The identity matrix
np.identity(5)
```

```python
#Repeat the array
arr= np.array([[1,2,3]])
r=np.repeat(arr,3,axis=0)
r
```

```python
#Print an given array 
a=np.ones((5,5),dtype='int')
z=np.zeros((3,3))
a[1:4,1:4]=z
a[2,2]=9
a
```

```python
#Copying an array
a= np.array([1,2,3])
b=a
b[0]=100
a 
```

```python
#In above example of copying an array the change in b also change the a. So be careful while copying the array (use copy()function)
a= np.array([1,2,3])
b=a.copy()
b[0]=100
a
```

### Reorganizing arrays

```python
before = np.array([[5,2,8],[7,4,1]])
print(before)
after= before.reshape((1,6))
print(after)
```

```python
#Vertically stacking vectors
v1= np.array([1,2,3,4])
v2=np.array([5,6,4,2])
np.vstack([v1,v2,v2])
```

```python
#Horizontal stacking vectors
v1= np.array([1,2,3,4])
v2=np.array([5,6,4,2])
np.hstack([v1,v2,v1])
```
