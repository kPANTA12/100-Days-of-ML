# First Neural Network (Classification)

#### 1. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

#### 2. Create Dataset

```python
# Hour Studies
X = np.array([1,2 ,3,4, 5, 6, 7, 8]).reshape(-1, 1)
#Pass (1) or Fail (0)
y= np.array([0, 0, 0, 0 , 1, 1 ,1 ,1 ])
```

#### 3.Built Neural Network

```python
model = keras.Sequential([
    keras.Input(shape=(2,1)),
    layers.Dense(8, activation = 'relu'),
    layers.Dense(1, activation = 'sigmoid')
])
```

#### 4. Compile Model

```python
model.compile(
    optimizer = 'adam',
    loss = 'binary_crossentropy',
    metrics =['accuracy']
)
```

#### 5. Train Model

```python
history = model.fit(X, y, epochs=500, verbose=0)
'''One epoch = the model sees the entire training dataset once.
If you have:
100 data points
Epoch = 1
The model uses all 100 points once to learn.'''
'''verbos -> Controls how much training information is printed.'''
```

#### 6. Evaluate

```python
loss , accuracy =model.evaluate(X, y)
print("Accuracy: " , accuracy)
```

#### 7. Prediction

```python
predictions = model.predict(np.array([[5], [7]], dtype=float))
print(predictions)
```
