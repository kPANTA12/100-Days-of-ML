# Recurrent Neural Networks (RNN)

#### 1. Import Libraries

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
```

#### 2. Create Simple Sequence Data

```python
X = np.array([
    [1, 2, 3, 4],
    [2, 3, 4, 5],
    [3, 4, 5, 6]
])

y = np.array([5, 6, 7])
```

#### 3. Reshape for RNN

```python
X = X.reshape((X.shape[0], X.shape[1], 1))
```

#### 4. Build RNN Model

```python
model = models.Sequential([
    layers.Input(shape=(4, 1)),
    layers.SimpleRNN(10, activation='tanh'),
    layers.Dense(1)
])
```

#### 5. Compile Model

```python
model.compile(
    optimizer='adam',
    loss='mse'
)
```

#### 6. Train Model

```python
model.fit(X, y, epochs=300, verbose=0)
```

#### 7. Test Prediction

```python
test = np.array([[4, 5, 6, 7]]).reshape((1, 4, 1))
prediction = model.predict(test)
print(prediction)
```
