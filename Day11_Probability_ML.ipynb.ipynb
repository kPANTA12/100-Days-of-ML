# Probability for Machine Learning 

## Load Necessary Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Coin Toss

```python
toss = np.random.choice(['H','T'],size=1000)
H= np.mean(toss=='H')
T = np.mean(toss == 'T')
print("Probability of heads =", H)
print("Probability of tails =", T)
```

## 2. Dice Roll

```python
roll = np.random.choice([1,2 , 3,4, 5,6], size= 100)
for i in range(1,7):
    print("Probability of ", i, " is = ", np.mean(roll==i))
```

## 3. Expected Value

```python
outcomes = np.array([1, 2, 3 ,4 ,5 ,6])
prob= 1/6
exp_value= np.sum(outcomes*prob)
print("Expected value:", exp_value)
```

## 4. Conditional Probability 

```python
# Simulate rain and umbrella
np.random.seed(0)
rain = np.random.choice([0, 1], size=10000, p=[0.7, 0.3])  # 30% chance of rain
umbrella = (rain == 1) | (np.random.rand(10000) < 0.1)     # 10% chance people carry umbrella unnecessarily

P_rain_given_umbrella = np.mean(rain[umbrella == 1] == 1)
print("P(Rain | Umbrella):", P_rain_given_umbrella)
```

## 5. Bayes Theorem

```python
# Medical test example
P_disease = 0.01
P_pos_given_disease = 0.95
P_pos_given_no_disease = 0.05
P_no_disease = 1 - P_disease

P_disease_given_pos = (P_pos_given_disease * P_disease) / (
    P_pos_given_disease * P_disease + P_pos_given_no_disease * P_no_disease
)

print("Probability of disease given positive test:", P_disease_given_pos)
```
