# Deep Learning Frameworks (TensorFlow & Keras)

#### 1. Import Libraries

```python
import numpy as np
import tensorflow as tf 
from tensorflow import keras
```

#### 2. Create Dataset

```python
# Inputs : hours Studied
x =np.array([1, 2, 3, 4, 5, 6], dtype=float).reshape(-1, 1)
# Output: Exam Score
y = np.array([35, 40, 50, 60, 70, 80], dtype=float)
```

#### 3. Build Model

```python
model = keras.Sequential([
    keras.Input(shape=(1,)),
    keras.layers.Dense(1)
])
```

#### 4. Compile Model

```python
model.compile(
    optimizer='sgd',
    loss='mean_squared_error'
)
```

#### 5. Train Model

```python
model.fit(x, y, epochs=500, verbose=0)
```

#### 6. Make Prediction

```python
model.predict(np.array([[7]]))
```
