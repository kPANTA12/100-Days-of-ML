# Advanced Feature Engineering 

#### 1. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectKBest, f_classif
```

#### 2. Load DataSet

```python
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target']= data.target
df.head()
```

#### 3. Feature Scaling

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df.drop('target',axis =1))
```

#### 4. Polynomial Feature Creation

```python
poly = PolynomialFeatures(degree = 2, include_bias =False)
X_poly = poly.fit_transform(X_scaled)
X_poly.shape
```

#### 5. Feature Selection (Select Best Features)

```python
selector = SelectKBest(score_func=f_classif, k=10)
X_selected = selector.fit_transform(X_scaled, df['target'])

X_selected.shape
```

#### 6. Train Model with Engineered Features

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_selected, df['target'], test_size=0.2, random_state=42
)

model = LogisticRegression(max_iter=5000)
model.fit(X_train, y_train)

print("Model Accuracy:", model.score(X_test, y_test))
```
