# Backpropagation & Gradient Descent

```python
import numpy as np
import matplotlib.pyplot as plt
def loss(w):
    return w**2
w = np.linspace(-5, 5, 100) # np.linspace(start, stop, num)
l = loss(w)
plt.plot(w, l)
plt.xlabel("Weight")
plt.ylabel("Loss")
plt.title("Loss vs Weight")
plt.show()
```

```python
w =4.0 #Initial Weight
lr = 0.1 # Learning rate
for step in range(10):
    grad = 2 * w        # derivative of w^2
    w = w - lr * grad
    print(f"Step {step+1}: Weight = {w}")
```

#### Some Note:
##### a. What is loss?
##### ->In a neural network, loss (or loss function) is a mathematical measure of how wrong the model's predictions are compared to the actual correct answers
##### b. What does gradient represent?
##### ->In a neural network, a gradient represents the rate of change of the network's error (loss) with respect to its internal parameters, such as weights and biases. 
##### c. Why subtract gradient?
##### -> In a neural network, we subtract the gradient because it points in the direction of the steepest increase in error (the "uphill" direction). Since the goal of training is to minimize error, we must move in the opposite direction—down the slope toward the minimum. 
##### d. What happens if learning rate is too high?
##### -> If the learning rate is too high, the neural network's training process becomes unstable and can fail entirely. Instead of making precise adjustments to minimize error, the model takes steps that are too large for the "terrain" of its loss function.
