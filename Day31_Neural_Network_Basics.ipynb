# Introduction to Neural Networks

```python
import numpy as np
x = np.array([5, 0.8]) #Inputs
w= np.array([0.6, 0.4]) #Weights
b = 0.2 # Bias
z= np.dot(x, w) +b
output = 1/ (1+np.exp(-z)) # Sigmoid activation
print("Neural Output: ", output)
```

#### Higher study hours → higher output

#### Higher weight → more influence

#### Output close to 1 → pass

#### Output close to 0 → fail
