# Feature Engineering Basic

## Setup (Use Titanic dataset)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('Titanic-Dataset.csv') 
df.head(5)
```

## Handling Missing Values

### a. Check Missing Values

```python
df.isnull().sum()
```

### b. Filling missing value numerical values

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df
```

### c. Filling Missing Cateforical Values

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
df
```

## Encoding Categorical Features (Convert text -> Numbers)

### a. Label encoding  

```python
df['Sex']=df['Sex'].map({'male':0,'female':1})
df
```

### b. One-Hot Encoding

```python
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)
df
```

## Feature Scaling (Normalize Numerical Feature)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df['Fare_scaled'] = scaler.fit_transform(df[['Fare']])
df
```

## Creating New Feature

### a. Family Size

```python
df['FamilySize'] = df['SibSp']+ df['Parch']+1
df
```

### b. Age Group

```python
df['AgeGroup']=pd.cut(df['Age'], bins=[0,12,18,60,100], labels= ['Child', 'Teen', 'Adult', 'Senior'])
df
```

## c. Fare Category


```python
df['FareCategory']= pd.qcut(df['Fare'], 4, labels=['Low', 'Medium', 'High', 'Very_High'])
df
```

### d. Title Extraction from Name

```python
df['Title'] = df['Name'].str.extract(' ([A-Za-z]+)\.')
df
```

## Removing Unnecessary Feature

```python
df.drop(['Ticket','Cabin','PassengerId','Name'], axis=1, inplace=True)
df
```

## Detecting Outliers (Boxplot)

```python
sns.boxplot(x=df['Fare'])
```

```python
q1 = df['Fare'].quantile(0.25)
q3 = df['Fare'].quantile(0.75)
iqr = q3 - q1

upper = q3 + 1.5 * iqr
lower = q1 - 1.5 * iqr

df = df[(df['Fare'] > lower) & (df['Fare'] < upper)]
```
