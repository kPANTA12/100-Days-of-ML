```python
#If statement in python
#find greater among two numbers
a= input("Enter first number:")
b=input("Enter second number:")
if a>b:
    print(a,"is greater than ",b)
else:
    print(b,"is greater than ",a)
    
```

```python
#Example for multiple options
if a>b:
    print(a,"is greater than ",b)
elif b>a:
    print(b,"is greater than ",a)
else:
    print("Both are equal.")
    
```

```python
#If condition in tuple
tup1=(1,2,3,4,5,6)
if 2 in tup1:
    print("2 is present in the tuple.")
else:
    print("2 is not present in the tuple.")
```

```python
#If with list
l1=[1,2,3,4,5,6]
if l1[1]==1:
    l1[1]=100+l1[1]
else:
    l1[1]=200+l1[1]
l1
    
```

```python
#If with dictionary
d1={"Apple":50,"Orange":30,"Cherry":90}
if d1["Apple"]==50:
    d1["Apple"]=d1["Apple"]+30
else:
    d1["Apple"]=d1["Apple"]+20
d1
```

```python
#Loops in python
#While loop(Program to write first 10 natural number)
i=1
while i<=10:
    print(i)
    i=i+1
```

```python
#multiplication table of 16
i=1
n=16
while i<=10:
    print(n," * ",i," = ", n*i)
    i=i+1
```

```python
#While loop in list(square all the element)
l1=[1,2,3,4,5]
i=0
while i<len(l1):
    l1[i]=l1[i]*l1[i]
    i=i+1
l1
```

```python
#For loop (print element of list using for loop)
l1=["Apple","Cherry","Orange"]
for item in l1:
    print(item)
```

```python
#Nested for loop
l1=["Apple","Orange","Banana"]
l2=["Programming","Mathematics","Physics","Electronics"]
for i in l1:
    for j in l2:
        print (i, j)
        
```

```python
#WAP to check even or odd
def even_odd(x):
    if x%2==0:
        print(x,"is even")
    else:
        print(x," is odd")
even_odd(15)
```

```python
#Lamba function
g= lambda x:x*x*x;
print(g(10))
```

```python
#lambda function with filter(print only odd elements)
l1=[10,15,81,65,12] 
final_list=list(filter(lambda x:x%2!=0,l1)) #Filter function filter the elements and have two arguments
print(final_list)
```

```python
#lambda function with map (update all element with square of element)
l1=[10,15,81,65,12] 
final_list=list(map(lambda x:x*x,l1)) #map function  update the elements and have two arguments
print(final_list)
```
